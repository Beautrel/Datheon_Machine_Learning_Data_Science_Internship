## Conclusions de l'exploration

**Qualité des données.** Le jeu contient 20 640 quartiers et 8 variables explicatives, toutes numériques, sans aucune valeur manquante : ni imputation ni encodage ne sont nécessaires.

**Échelles et valeurs extrêmes.** Les variables ont des échelles très différentes (`MedInc` va jusqu'à 15, `Population` jusqu'à 35 682), ce qui impose une normalisation, surtout pour Ridge et Lasso. Certaines valeurs sont aberrantes, par exemple `AveOccup` dont le maximum atteint 1243 alors que 75 % des quartiers sont sous 3,3.

**Cible plafonnée.** Le prix médian `MedHouseVal` est plafonné à 5,0 (500 000 $) : environ 4,80 % des quartiers y sont empilés. Le modèle n'a jamais vu de prix au-dessus, il sous-estimera donc les quartiers les plus chers.

**Corrélations.** `MedInc` est la seule variable fortement liée au prix (0,69). Deux paires de variables explicatives sont fortement corrélées entre elles : `AveRooms`/`AveBedrms` (0,85) et `Latitude`/`Longitude` (−0,92). Cette multicolinéarité rend l'OLS instable et justifie la comparaison avec Ridge et Lasso.
